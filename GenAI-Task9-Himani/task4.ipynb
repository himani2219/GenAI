```python
import numpy as np
```

```python
data = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])
```

```python
data
```

```python
# row wise sum
row_sum = np.sum(data, axis=1)
row_sum
```

```python
# column wise sum
col_sum = np.sum(data, axis=0)
col_sum
```

```python
# minimum value
min_value = np.min(data)
min_value
```

```python
# maximum value
max_value = np.max(data)
max_value
```

```python
# overall mean
overall_mean = np.mean(data)
overall_mean
```
