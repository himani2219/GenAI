```python
import numpy as np
marks = np.array([78, 85, 90, 66, 72, 88, 95, 60])
```

```python
np.sort(marks)
```

```python
print(f"90 percentile: {np.percentile(marks, 90)}")
print(f"75 percentile: {np.percentile(marks, 75)}")
print(f"50 percentile: {np.percentile(marks, 50)}")
```

```python
avg = np.mean(marks)
print(marks)
print(f"Average: {avg}")
# approach 1 = wanted to filter the array as we do in js. so used lambda function
print(len(list(filter(lambda x: x > avg, marks))))
#approach 2 = searched for a shorter way and found that we can directly filter the array using boolean indexing
print(len(marks[marks > avg]))
```
