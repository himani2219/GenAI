```python
import numpy as np
sales = np.array([1200, 1500, 900, 2000, 1800, 1700, 1600])
```

```python
print(f"Total weekly sales: {np.sum(sales)}")
```

```python
print(f"Average daily sales: {np.mean(sales)}")
```

```python
print(f"Highest daily sales: {np.max(sales)}")
print(f"Lowest daily sales: {np.min(sales)}")
```

```python
print(f"Standard deviation of sales: {np.std(sales)}")
```

```python
# used list comprehension  
print(f"Days when sales were above average: {np.array([x for x in sales if x > np.mean(sales)])}")
```
