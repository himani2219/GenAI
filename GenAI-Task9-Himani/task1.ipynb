```python
import numpy as np
arr1 = np.arange(1,11).astype(int)
arr1
```

```python
arr2 = np.arange(1,10).reshape(3,3)
arr2
```

```python
list1 = [10, 20, 30, 40, 50]
arr3 = np.array(list1)
arr3
```

```python
arrays = {
    "arr1": arr1,
    "arr2": arr2,
    "arr3": arr3
}
for name, arr in arrays.items():
    print(f"data type of {name} is: {(arr).dtype}")
    print(f"shape of {name} is: {(arr).shape}")
```
