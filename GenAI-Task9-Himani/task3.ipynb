```python
import numpy as np
```

```python
values = np.array([2, 4, 6, 8, 10])
print(f"square root of values: {np.sqrt(values)}")
print(f"exponential of values: {np.exp(values)}")
print(f"natural logarithm of values: {np.log(values)}")
print(f"sum of values: {np.sum(values)}")
print(f"cumulative sum of values: {np.cumsum(values)}")
```
